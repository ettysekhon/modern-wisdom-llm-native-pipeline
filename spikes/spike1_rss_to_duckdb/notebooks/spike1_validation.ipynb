```python
from pathlib import Path

import duckdb
from IPython.display import display

repo_root = (
    Path(__file__).resolve().parents[2] if "__file__" in globals() else Path.cwd().parents[2]
)
db_path = repo_root/"data"/"duckdb"/"modern_wisdom.duckdb"

con = duckdb.connect(str(db_path), read_only=True)

tables = con.execute("""
    SELECT *
    FROM information_schema.tables
""").df()

episodes = con.execute("""
    SELECT *
    FROM mw.episodes
    ORDER BY publish_date DESC
""").df()

per_year = con.execute("""
    SELECT strftime(CAST(publish_date AS DATE), '%Y') AS year,
           COUNT(*) AS count
    FROM mw.episodes
    WHERE publish_date IS NOT NULL
    GROUP BY year
    ORDER BY year
""").df()

display(tables)
display(episodes)
per_year
```
