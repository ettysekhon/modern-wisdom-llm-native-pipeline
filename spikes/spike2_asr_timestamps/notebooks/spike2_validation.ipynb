```python
from pathlib import Path

import pandas as pd

repo_root = (
    Path(__file__).resolve().parents[2] if "__file__" in globals() else Path.cwd().parents[2]
)
TRANSCRIPTS_DIR = repo_root / "data" / "transcripts"


episode_id = "322ad7ce-780e-4aa5-af07-9a3828e75e94"
parquet_file = TRANSCRIPTS_DIR / f"episode_id={episode_id}" / "part-00000.snappy.parquet"

df = pd.read_parquet(parquet_file)

print("Columns:", df.columns.tolist())
print("\nFirst 5 rows:")
display(df.head())

# Combine into a readable transcript preview
print("\nTranscript Preview:")
for _, row in df.head(10).iterrows():
    speaker = f"[{row.speaker}]" if "speaker" in df.columns and pd.notna(row.get("speaker")) else ""
    print(f"{speaker} ({row.start_ts:.1f}–{row.end_ts:.1f}s): {row.text}")

# Full text (joined)
full_text = " ".join(df["text"].astype(str).tolist())
print("\n--- FULL TEXT (first 1000 chars) ---")
print(full_text[:1000] + ("..." if len(full_text) > 1000 else ""))
```
