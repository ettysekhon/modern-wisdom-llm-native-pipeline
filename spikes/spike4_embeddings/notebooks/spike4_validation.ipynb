```python
from pathlib import Path

import duckdb
import pandas as pd
from IPython.display import display

pd.set_option("display.max_rows", None)

repo_root = (
    Path(__file__).resolve().parents[2] if "__file__" in globals() else Path.cwd().parents[2]
)
db_path = repo_root / "data" / "duckdb" / "modern_wisdom.duckdb"

# context manager (auto-closes even on errors)
with duckdb.connect(str(db_path), read_only=True) as con:
    tables = con.execute("""
    SELECT emb_v, model_id, provider, COUNT(*) AS rows, MIN(created_at), MAX(created_at)
    FROM embeddings
    WHERE episode_id = '0a4fa77e-bc0f-11ef-bab6-3f37b4906b43'
    GROUP BY 1,2,3;
""").df()

display(tables)
```
